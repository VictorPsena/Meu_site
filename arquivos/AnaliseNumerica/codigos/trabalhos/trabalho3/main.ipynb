```python
"""
1/3 de Simpson reptidada para integrais duplas
"""
def f(x,y):
    return x + y
def int_simpson(num_par_1: int, num_par_2: int ,int_1 = [float, float], int_2 = [float, float]):
    n1, n2 = num_par_1, num_par_2
    
    if n1%2 != 0 or n2%2 != 0:
        return print('O intervalo precisa ser subdividido em um número par')

    x = np.linspace(int_1[0], int_1[1], num_par_1 + 1)  
    y = np.linspace(int_2[0], int_2[1], num_par_2 + 1)

    h1 = (x[1] - x[0])
    h2 = (y[1] - y[0])
    
    ## Calculando a Aproximação pela regra 1/3 de simpson reptida
    soma_1 = sum([f(x[2*i], y[0]) for i in range(int(n1/2 - 1))])
    soma_2 = sum([f(x[2*i-1], y[0]) for i in range(1, int(n1/2))])

    soma_3 = sum([f(x[0], y[2*i]) for i in range(int(n2/2 - 1))])
    soma_4 = sum([sum([f(x[2*i], y[2*j]) for i in range(int(n1/2 - 1))]) for j in range(int(n2/2 - 1))])
    soma_5 = sum([sum([f(x[2*i - 1], y[2*j]) for i in range(1, int(n1/2))]) for j in range(int(n2/2 - 1))])
    soma_6 = sum([f(x[-1], y[2*i]) for i in range(int(n2/2 - 1))])

    soma_7 = sum([f(x[0], y[2*i - 1]) for i in range(1, int(n2/2))])
    soma_8 = sum([sum([f(x[2*i], y[2*j - 1]) for i in range(int(n1/2 - 1))]) for j in range(1, int(n2/2))])
    soma_9 = sum([sum([f(x[2*i - 1], y[2*j - 1]) for i in range(1, int(n1/2))]) for j in range(1, int(n2/2))])
    soma_10 = sum([f(x[-1], y[2*i - 1]) for i in range(1, int(n2/2))])

    soma_11 = sum([f(x[2*i], y[-1]) for i in range(int(n2/2 - 1))])
    soma_12 = sum([f(x[2*i-1], y[-1]) for i in range(1, int(n2/2))])

    # Montando a Expressão
    a = (h1*h2)/9
    int_x = a * ((f(x[0], y[0]) + 2 * soma_1 + 4 * soma_2) + 2 * (soma_3 + 2 * soma_4 + 4 * soma_5 + soma_6) + 4 * (soma_7 + 2 * soma_8 + 4 * soma_9
                + soma_10) + (f(x[0], y[-1]) + 2 * soma_11 + 4 * soma_12 + f(x[-1], y[-1])))
    return int_x
```

```python
import numpy as np

def f(x, y):
    return np.log(x + 2*y)

def int_simpson_dupla(f, a, b, c, d, n, m):
    if n % 2 != 0 or m % 2 != 0:
        raise ValueError("n e m precisam ser pares")

    x = np.linspace(a, b, n + 1)
    y = np.linspace(c, d, m + 1)
    hx = (b - a) / n
    hy = (d - c) / m

    S = 0.0
    for i in range(n + 1):
        wx = 1 if i in (0, n) else (4 if i % 2 else 2)
        for j in range(m + 1):
            wy = 1 if j in (0, m) else (4 if j % 2 else 2)
            S += wx * wy * f(x[i], y[j])

    return (hx * hy / 9.0) * S
```
